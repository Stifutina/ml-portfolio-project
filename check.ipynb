```python
import pandas as pd

data = pd.DataFrame({
    "message": [
        "Здравствуйте, не могу оплатить заказ, ошибка на странице оплаты",
        "Когда будет доставка моего заказа №4521?",
        "Хочу вернуть товар, он оказался бракованным",
        "Спасибо большое, всё отлично, доставили быстро!",
    ],
    "intent": ["billing_issue", "delivery_status", "return_request", "positive_feedback"],
})

print(data)
print(f"\nВсего примеров: {len(data)}")
print(f"Уникальных классов: {data['intent'].nunique()}")
```
